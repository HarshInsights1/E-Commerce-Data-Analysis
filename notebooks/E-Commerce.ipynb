```python
# Step 1 : Import required Libraries
import pandas as pd 
import numpy as np
```

```python
# Load Dataset 
df = pd.read_excel('Ecommerce_Unclean_Project.xlsx')
```

```python
# Remove Duplicates 
df.drop_duplicates(inplace = True)
```

```python
df.duplicated().sum()
```

```python
# step 6 : Replace Invalid Values 
# step 7 : Replace common invalid values with NaN
df.replace(['N/A','NULL',''],np.nan,inplace = True)
```

```python
# Remove Extra Spaces 
df = df.apply(lambda x: x.str.strip() if x.dtype == 'object' else x)
```

```python
# Standaridize Text Format
# Convert text into power case
df['Customer_Name'] = df['Customer_Name'].str.title()
df['City'] = df['City'].str.title()
df['State'] = df['State'].str.title()
    
```

```python
# Validate Email Address
df = df[df['Email'].str.contains('@',na = False)]
```

```python
# Convert Date Column type
df['Order_Date'] = pd.to_datetime(df['Order_Date'],errors = 'coerce') # NAT
df['Delivery_Date'] = pd.to_datetime(df['Delivery_Date'], errors = 'coerce')
```

```python
# Convert Numeric Columns
cols = ['Qty','Unit_Price','Discount']
for c in cols :
    df[c] = pd.to_numeric(df[c], errors = 'coerce')
```

```python
# Remove Invalid Quantity
# Remove records having -ve or 0
df= df[df['Qty']>0]
```

```python
# Fill Missing Values
df['Discount'] = df['Discount'].fillna(0)
df['Phone'] = df['Phone'].fillna('Unknown')
df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
# Feature Engineering 
# Add Total Sales Column
df['Sales'] = df['Qty'] * df['Unit_Price']
```

```python
df.head(5)
```

```python
# Net Amount 
df['Net_Amount'] = df['Sales'] - (df['Sales']*df['Discount']/100)
```

```python
df
```

```python
df['Profit'] = df['Net_Amount'] * 0.20
```

```python
df['Month'] = df['Order_Date'].dt.month_name()
```

```python
df
```

```python
df['year'] = df['Order_Date'].dt.year
```

```python
df
```

```python
df['Weekday'] = df['Order_Date'].dt.day_name()
```

```python
df = df.drop(columns='Weekday')
```

```python
# Import Data into MySql
from sqlalchemy import create_engine, text

engine = create_engine(
    "mssql+pyodbc://@localhost/ecommerce?"
    "trusted_connection=yes&"
    "driver=ODBC+Driver+18+for+SQL+Server&"
    "Encrypt=Optional&"
    "TrustServerCertificate=yes"
)

with engine.connect() as connection:
    result = connection.execute(text("""
        SELECT
            @@SERVERNAME AS server_name,
            DB_NAME() AS database_name,
            SYSTEM_USER AS login_name
    """))
    print(result.fetchone())
```

```python
df.to_sql(
    name='orders',
    con=engine,
    schema='dbo',
    if_exists='replace',
    index=False
)
```
